Q26

```python
sales = pd.Series([450, 620, 300, 780], index=['TV', 'Laptop', 'Microwave', 'Fridge'])
print(sales.values)
print(sales.index)
print(sales['Laptop'])
print(sales[sales > 500])
print("Does 'TV' exist in index?", 'TV' in sales)
sales_dict = sales.to_dict()
print(sales_dict)
```

Q28

```python
sales = pd.Series([450, 720, 310, 890], index=['Pen', 'Notebook', 'Eraser', 'Backpack'])
print("Values:", sales.values)
print("Index:", sales.index)
print("\nSales for Notebook:", sales['Notebook'])
print("\nSales above 500:\n", sales[sales > 500])
print("\nDoes 'Pen' exist?", 'Pen' in sales)
sales_dict = sales.to_dict()
print("\nConverted Dictionary:", sales_dict)
```
